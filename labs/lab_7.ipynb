```python
#Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics
```

```python
data = pd.read_csv("../data/secondary_data.csv", sep = ";")
data.head()
```

Can we predict the class of mushroom? 

```python
print(data.dtypes)
```

```python
#Dropping our columns with missing values
mushroom_data = data.dropna(axis = 1)
mushroom_data.head()
```

Going to make class, does-bruise-or-bleed, has-ring, and season as categorical

```python
categorical_cols = ['class', 'does-bruise-or-bleed', 'has-ring', 'season']
mushroom_data[categorical_cols] = mushroom_data[categorical_cols].astype("category")
print(mushroom_data.dtypes)
```

```python
# Standardizing our numeric variables
mushroom_data[["stem-height","cap-diameter","stem-width"]]= StandardScaler().fit_transform(mushroom_data[["stem-height","cap-diameter","stem-width"]])
```

```python
# Only want to keep categorical and numerical, don't want to keep objects
mushroom_data = mushroom_data.loc[:, ['class', 'does-bruise-or-bleed', 'has-ring', 'season', 'stem-height', 'cap-diameter', 'stem-width']]
mushroom_data.head()
```

```python
mushroom_data_1h = pd.get_dummies(mushroom_data, columns=['class', 'does-bruise-or-bleed', 'has-ring', 'season'], drop_first=True)
mushroom_data_1h.head()
```

```python
# Defining our independent/dependent variables
X = mushroom_data_1h.drop('class_p', axis=1)
y = mushroom_data_1h['class_p']   
```

```python
# Train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.70, stratify = y, random_state=77)
```

```python
# Model building
shroom_nn = KNeighborsClassifier(n_neighbors=7)
shroom_nn.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_auc_score, roc_curve
import warnings

accuracy = accuracy_score(y_test, shroom_nn.predict(X_test))

tn, fp, fn, tp = confusion_matrix(y_test, shroom_nn.predict(X_test)).ravel()
tpr = tp / (tp + fn)
fpr = fp / (fp + tn)

# Calculate F1 score
f1 = metrics.f1_score(y_test, shroom_nn.predict(X_test))

# Calculate cross entropy
cross_entropy = -np.mean(y_test * np.log(shroom_nn.predict_proba(X_test)[:, 1]) + (1 - y_test) * np.log(1 - shroom_nn.predict_proba(X_test)[:, 1]))

# Calculate ROC/AUC
roc_auc = roc_auc_score(y_test, shroom_nn.predict_proba(X_test)[:, 1])
fpr, tpr, thresholds = roc_curve(y_test, shroom_nn.predict_proba(X_test)[:, 1])

print('Accuracy:', accuracy)
print('TPR:', tpr)
print('FPR:', fpr)
print('F1 Score:', f1)
print('Cross Entropy:', cross_entropy)
print('ROC AUC:', roc_auc)
warnings.filterwarnings("ignore")
```
